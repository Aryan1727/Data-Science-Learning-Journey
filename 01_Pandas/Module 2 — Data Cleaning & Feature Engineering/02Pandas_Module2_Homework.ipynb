# Part C — Feature Engineering
### *Main objective: To create useful features for further analysis and machine learning.*

`````python
# Create FamilySize column.
df["FamilySize"] = df["Parch"] + df["SibSp"] + 1
display(df['FamilySize'].head(5))
`````

`````python
# Create AgeGroup column.
df['AgeGroup'] = pd.cut(
    df['Age'],
    bins = [0, 18, 60, 100],
    labels= ["Child", "Adult", "Seniore"]
)
display(df["AgeGroup"].head(5))
`````

`````python
# Create categories basis on FamilySize:
# 1 → Alone
# 2–4 → Small Family
# 5+ → Large Family

df["Category_FamilySize"] = pd.cut(
    df['FamilySize'],
    bins = [0, 1, 4, float("inf")],
    labels = ["Alone", "Small Family", "Large Family"]
)

display(df['Category_FamilySize'].head())
`````

*After trying this, I understood the concept clearly. If NumPy is also being used, I can write `np.inf` instead of `float("inf")`.*
