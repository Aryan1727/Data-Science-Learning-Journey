# 3. How to **Handle Missing Values**

### *Method  — fillna()*
*Fill Age column NaN values by its median because Age is a numerical column.*


### Missing Values Handling
**We use different methods depending on the type of data:**

*- Numeric column: median/mean*

*- Categorical column: mode*

*- If many values are missing, use a custom label like "Missing" or "Unknown"*


```python

# 1) Age: numeric column
df["Age"] = df["Age"].fillna(df["Age"].median())

# 2) Embarked: categorical column
mode_value = df["Embarked"].mode()
if not mode_value.empty:
    df["Embarked"] = df["Embarked"].fillna(mode_value.iloc[0])
else:
    df["Embarked"] = df["Embarked"].fillna("Unknown")

# 3) Cabin: many missing values, so use custom label
df["Cabin"] = df["Cabin"].fillna("Value_Missing")
```

```python
# Check null values again for cheking function is properly works or not
df.isnull().sum()
```

# 9A .How to use **Single_groupby()**
### *This is an important Pandas concept: finding the average fare for male and female passengers.*

```python
df.groupby("Sex", observed=False)["Fare"].mean()
```

### Logic Behind `groupby()`

Create groups based on `Sex`  
↓  
Male  
Female  
↓  
Select the `Fare` values from each group  
↓  
Calculate the average fare

# 9B. How to use **Multiple Groupby**


```python
df.groupby(["Sex", "Pclass"], observed=False)["Fare"].mean()
```

# 10. How to use **Miltiple Aggregations**
agar Average age , Maximum Age, Average Fare

```python
df.groupby("Sex", observed=False).agg({
    "Age": "mean",
    "Fare": "mean"
})
```

```python
df.groupby("Sex", observed=False).agg({
    "Age": ["mean", "max"],
    "Fare": ["mean", "max"]
})
```

### My Self-Learning About `groupby()`

***Through self-learning, I discovered that the `groupby()` method treats object and categorical data types differently. When a categorical column is used, the code runs successfully but may display a `FutureWarning`:**,*

***`The default of observed=False is deprecated and will be changed to True in a future version of pandas.`**,*

***To remove this warning, I asked the AI assistant in VS Code for help. I learned how Pandas handles object and categorical data types differently when using `groupby()`.**,*
***At first, I thought about changing the data type of the `Sex` column. However, I learned that adding `observed=False` makes the code explicit:**,*

***`df.groupby("Sex", observed=False)`**,*
***This removes the warning while allowing me to continue using the efficiency of the categorical data type.**,*

### **This was my self-learning about `groupby()`.**
